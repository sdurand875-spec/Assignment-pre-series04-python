# 4. Estimate the tangential speed of the signal

## 4.1 Tangential velocity vs tangential speed

- **Tangential velocity** is a **vector**: an arrow with a direction and a length. It tells us "I am going this way, at this rate." At each moment, this vector is always **tangent to the path** of the movement, which is why we call it *tangential* velocity.
- **Tangential speed** is a **scalar**: just a number, the length of that arrow. It only tells us "I am going at this rate," without any direction.

In other words, the tangential speed is the **magnitude** (or norm) of the tangential velocity vector:

$$\text{speed} = \|\vec{v}\|$$

This link is what we use in the next section to compute the tangential speed.

## 4.2 Tangential speed of a 3D movement

For a movement in 3D, the position at time $t$ is given by $x(t)$, $y(t)$ and $z(t)$. Their time derivatives $\dot{x}(t)$, $\dot{y}(t)$ and $\dot{z}(t)$ are the components of the velocity vector along each axis (computed in section 3):

$$\vec{v}(t) = \big(\dot{x}(t),\ \dot{y}(t),\ \dot{z}(t)\big)$$

The tangential speed is the magnitude (norm) of this vector. Using the Pythagorean theorem in 3D, we get:

$$v(t) = \|\vec{v}(t)\| = \sqrt{\dot{x}(t)^2 + \dot{y}(t)^2 + \dot{z}(t)^2}$$

The result is one positive number at each time $t$.

**Reference:** Ling, S. J., Sanny, J., & Moebs, W. (2016). *University Physics Volume 1*. OpenStax, section 4.1, Displacement and Velocity Vectors: https://openstax.org/books/university-physics-volume-1/pages/4-1-displacement-and-velocity-vectors

## 4.3 Tangential speed of an nD movement

The formula from 4.2 works in 3D, but we want a formula that works for **any number of dimensions** (1D, 2D, 3D... nD). Here is how we get it, step by step.

**Step 1 — Start from what we know (Pythagorean theorem).**
In 2D, a vector with components $(\dot{x}, \dot{y})$ is the hypotenuse of a right triangle, so its length is:
$$v = \sqrt{\dot{x}^2 + \dot{y}^2}$$
In 3D, we simply add one more squared term (see 4.2):
$$v = \sqrt{\dot{x}^2 + \dot{y}^2 + \dot{z}^2}$$

**Step 2 — Find the pattern.**
Each time we add a dimension, we add one more squared component under the square root.

**Step 3 — Number the axes instead of using letters.**
With many dimensions, we do not have enough letters ($x, y, z$...). So we call the position on axis $i$: $s_i(t)$, with $i = 1, 2, \dots, n$. Its time derivative $\dot{s}_i(t)$ is the velocity component along axis $i$. The 3D formula becomes:
$$v = \sqrt{\dot{s}_1^2 + \dot{s}_2^2 + \dot{s}_3^2}$$
and for $n$ dimensions:
$$v = \sqrt{\dot{s}_1^2 + \dot{s}_2^2 + \dots + \dot{s}_n^2}$$

**Step 4 — How to read the sum symbol $\Sigma$, and why there is a square root.**

*Reading the sum:* the symbol $\Sigma$ (the Greek capital letter "sigma") means "sum". It is read in three parts:
- **below** the $\Sigma$: where the counter starts, here $i = 1$;
- **above** the $\Sigma$: where the counter stops, here $n$;
- **to the right** of the $\Sigma$: what we add at each step, here $\dot{s}_i^2$.

So $\displaystyle\sum_{i=1}^{n} \dot{s}_i^2$ is read: *"the sum, for $i$ going from 1 to $n$, of $\dot{s}_i$ squared"*. It works like a `for` loop in Python: we start at $i = 1$, add $\dot{s}_1^2$, then $\dot{s}_2^2$, and so on until $\dot{s}_n^2$. For example, with $n = 3$:
$$\sum_{i=1}^{3} \dot{s}_i^2 = \dot{s}_1^2 + \dot{s}_2^2 + \dot{s}_3^2$$

**Step 5 — Write it in a compact way with the sum symbol $\Sigma$.**
$\Sigma$ means "add up all the terms", like a `for` loop: for $i$ going from 1 to $n$, we add $\dot{s}_i^2$. This gives the general formula:
$$v(t) = \sqrt{\sum_{i=1}^{n} \dot{s}_i(t)^2}$$
This is the **Euclidean norm** of the velocity vector.

**Step 6 — Check the formula with simple cases.**
- If $n = 3$, we find again the 3D formula from 4.2. ✔
- If $n = 1$, we get $v = \sqrt{\dot{s}_1^2} = |\dot{s}_1|$, i.e. the **absolute value** of the velocity. For example, if the velocity is $-3$, the speed is $3$. This makes sense: speed is a norm, so it is always positive or zero. ✔

**Reference:** Euclidean norm, in *Norm (mathematics)*, Wikipedia: https://en.wikipedia.org/wiki/Norm_(mathematics)#Euclidean_norm
