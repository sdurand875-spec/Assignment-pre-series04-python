# 4. Estimate the tangential speed of the signal

## 4.1 Tangential velocity vs tangential speed

- **Tangential velocity** is a **vector**: an arrow with a direction and a length. It tells us "I am going this way, at this rate." At each moment, this vector is always **tangent to the path** of the movement, which is why we call it *tangential* velocity.
- **Tangential speed** is a **scalar**: just a number, the length of that arrow. It only tells us "I am going at this rate," without any direction.

In other words, the tangential speed is the **magnitude** (or norm) of the tangential velocity vector:

$$\text{speed} = \|\vec{v}\|$$

This link is what we use in the next section to compute the tangential speed.

**Reference:** Ling, S. J., Sanny, J., & Moebs, W. (2016). *University Physics Volume 1*. OpenStax.
- Section 3.2, Instantaneous Velocity and Speed: https://openstax.org/books/university-physics-volume-1/pages/3-2-instantaneous-velocity-and-speed
- Section 4.1, Displacement and Velocity Vectors: https://openstax.org/books/university-physics-volume-1/pages/4-1-displacement-and-velocity-vectors

## 4.2 Tangential speed of a 3D movement

For a movement in 3D, the position at time $t$ is given by $x(t)$, $y(t)$ and $z(t)$. Their time derivatives $\dot{x}(t)$, $\dot{y}(t)$ and $\dot{z}(t)$ are the components of the velocity vector along each axis (computed in section 3):

$$\vec{v}(t) = \big(\dot{x}(t),\ \dot{y}(t),\ \dot{z}(t)\big)$$

The tangential speed is the magnitude (norm) of this vector. Using the Pythagorean theorem in 3D, we get:

$$v(t) = \|\vec{v}(t)\| = \sqrt{\dot{x}(t)^2 + \dot{y}(t)^2 + \dot{z}(t)^2}$$

The result is one positive number at each time $t$.

**Reference:** Ling, S. J., Sanny, J., & Moebs, W. (2016). *University Physics Volume 1*. OpenStax, section 4.1, Displacement and Velocity Vectors: https://openstax.org/books/university-physics-volume-1/pages/4-1-displacement-and-velocity-vectors

## 4.3 Tangential speed of an nD movement

The formula from 4.2 works in 3D, but we want a formula that works for **any number of dimensions** (1D, 2D, 3D... nD). Here is how we get it, step by step.

**Step 1 — Start from what we know (Pythagorean theorem).**
In 2D, a vector with components $(\dot{x}, \dot{y})$ is the hypotenuse of a right triangle, so its length is:
$$v = \sqrt{\dot{x}^2 + \dot{y}^2}$$
In 3D, we simply add one more squared term (see 4.2):
$$v = \sqrt{\dot{x}^2 + \dot{y}^2 + \dot{z}^2}$$

**Step 2 — Find the pattern.**
Each time we add a dimension, we add one more squared component under the square root.

**Step 3 — Number the axes instead of using letters.**
With many dimensions, we do not have enough letters ($x, y, z$...). So we call the position on axis $i$: $s_i(t)$, with $i = 1, 2, \dots, n$. Its time derivative $\dot{s}_i(t)$ is the velocity component along axis $i$. The 3D formula becomes:
$$v = \sqrt{\dot{s}_1^2 + \dot{s}_2^2 + \dot{s}_3^2}$$
and for $n$ dimensions:
$$v = \sqrt{\dot{s}_1^2 + \dot{s}_2^2 + \dots + \dot{s}_n^2}$$

**Step 4 — How to read the sum symbol $\Sigma$, and why there is a square root.**

*Reading the sum:* the symbol $\Sigma$ (the Greek capital letter "sigma") means "sum". It is read in three parts:
- **below** the $\Sigma$: where the counter starts, here $i = 1$;
- **above** the $\Sigma$: where the counter stops, here $n$;
- **to the right** of the $\Sigma$: what we add at each step, here $\dot{s}_i^2$.

So $\displaystyle\sum_{i=1}^{n} \dot{s}_i^2$ is read: *"the sum, for $i$ going from 1 to $n$, of $\dot{s}_i$ squared"*. It works like a `for` loop in Python: we start at $i = 1$, add $\dot{s}_1^2$, then $\dot{s}_2^2$, and so on until $\dot{s}_n^2$. For example, with $n = 3$:
$$\sum_{i=1}^{3} \dot{s}_i^2 = \dot{s}_1^2 + \dot{s}_2^2 + \dot{s}_3^2$$

**Step 5 — Write it in a compact way with the sum symbol $\Sigma$.**
$\Sigma$ means "add up all the terms", like a `for` loop: for $i$ going from 1 to $n$, we add $\dot{s}_i^2$. This gives the general formula:
$$v(t) = \sqrt{\sum_{i=1}^{n} \dot{s}_i(t)^2}$$
This is the **Euclidean norm** of the velocity vector.

**Step 6 — Check the formula with simple cases.**
- If $n = 3$, we find again the 3D formula from 4.2. ✔
- If $n = 1$, we get $v = \sqrt{\dot{s}_1^2} = |\dot{s}_1|$, i.e. the **absolute value** of the velocity. For example, if the velocity is $-3$, the speed is $3$. This makes sense: speed is a norm, so it is always positive or zero. ✔

**Reference:** Euclidean norm, in *Norm (mathematics)*, Wikipedia: https://en.wikipedia.org/wiki/Norm_(mathematics)#Euclidean_norm

## 4.4 Function to compute the tangential speed

The function follows the formula from 4.3 in four steps:
1. compute the time derivative of the signal (with the `central_difference` function from section 3);
2. square each component;
3. add the squared components over the dimensions (the $\Sigma$), for each sample;
4. take the square root.

**Input order:** `(signal, time)`. The signal is the main data we work on, and time only gives the date of each sample. We use the same order as the derivative functions in section 3 (and as `numpy.gradient(signal, time)`), so that all the functions of the notebook are used in the same way.

**Output shape:** `(samples, 1)`. The speed is a scalar, so there is only **one value per sample**, whatever the number of dimensions of the signal. For example, if `signal` has the shape `(100, 3)`, the output has the shape `(100, 1)`. We keep a column vector (with `keepdims=True`), like `time`, so that the speed can easily be plotted against time or compared with other column vectors.

```python
def tangential_speed(signal, time):
    """Tangential speed of a movement in nD.
    signal: array (samples, dimensions); time: array (samples, 1).
    Returns an array (samples, 1): the speed at each sample."""
    signal_derive = central_difference(signal, time)                 # 1. time derivative
    signal_squared = signal_derive ** 2                              # 2. square each component
    signal_add = np.sum(signal_squared, axis=1, keepdims=True)       # 3. sum over dimensions (Σ)
    speed = np.sqrt(signal_add)                                      # 4. square root
    return speed
```

## 4.5 Test cell

To check that `tangential_speed` works for any number of dimensions, we test it on **known signals**: movements in a straight line at constant velocity, from 1D to 4D. For each one, we first compute the expected speed **by hand** with the formula from 4.3:

$$v = \sqrt{\sum_{i=1}^{n} \dot{s}_i^2}$$

For a position $s_i(t) = a_i \, t$, the velocity component is simply $\dot{s}_i = a_i$, so the speed is constant.

| Test | Movement | Velocity components | Expected speed |
|---|---|---|---|
| 1D | $x = 2t$ | $2$ | $\sqrt{4} = 2$ |
| 1D (negative velocity) | $x = -3t$ | $-3$ | $\sqrt{9} = 3$ |
| 2D | $x = 3t,\ y = 4t$ | $3,\ 4$ | $\sqrt{9 + 16} = 5$ |
| 3D | $x = t,\ y = 2t,\ z = 2t$ | $1,\ 2,\ 2$ | $\sqrt{1 + 4 + 4} = 3$ |
| 4D | $s_i = t,\ 2t,\ 2t,\ 4t$ | $1,\ 2,\ 2,\ 4$ | $\sqrt{1 + 4 + 4 + 16} = 5$ |

We use straight lines on purpose: for a linear signal, the central difference gives the **exact** derivative, so the function must return exactly the expected speed. We compare the results with `np.allclose`, which tolerates the very small rounding errors of the computer. If a test fails, `assert` stops the code with an `AssertionError`.

```python
# Time vector for all tests: 50 samples from 0 to 1 s, as a column vector
time_test = np.linspace(0, 1, 50).reshape(-1, 1)

# --- Test 1D: x = 2t, expected speed = 2 ---
signal_1d = 2 * time_test
print("signal_1d.shape =", signal_1d.shape)   # expected: (50, 1)

speed_1d = tangential_speed(signal_1d, time_test)
print("speed_1d.shape =", speed_1d.shape)     # expected: (50, 1)
print("speed_1d =", speed_1d)                 # expected: [[2.], [2.], ..., [2.]]  
assert np.allclose(speed_1d, 2)

# --- Test 1D negative: x = -3t, expected speed = 3 ---
signal_1d_neg = -3 * time_test
speed_1d_neg = tangential_speed(signal_1d_neg, time_test)
print("speed_1d_neg =", speed_1d_neg)         # expected: [[3.], [3.], ..., [3.]]
assert np.allclose(speed_1d_neg, 3)

# --- Test 2D: x = 3t, y = 4t, expected speed = sqrt(3^2 + 4^2) = 5 ---
signal_2d = np.column_stack([3 * time_test, 4 * time_test])
speed_2d = tangential_speed(signal_2d, time_test)
print("speed_2d =", speed_2d)                 # expected: [[5.], [5.], ..., [5.]]
assert np.allclose(speed_2d, 5)

# --- Test 3D: x = t, y = 2t, z = 2t, expected speed = sqrt(1^2 + 2^2 + 2^2) = 3 ---
signal_3d = np.column_stack([1 * time_test, 2 * time_test, 2 * time_test])
speed_3d = tangential_speed(signal_3d, time_test)
print("speed_3d =", speed_3d)                 # expected: [[3.], [3.], ..., [3.]]
assert np.allclose(speed_3d, 3) 

# --- Test 4D: x = t, y = 2t, z = 2t, w = 4t, expected speed = sqrt(1^2 + 2^2 + 2^2 + 4^2) = sqrt(25) = 5 ---
signal_4d = np.column_stack([1 * time_test, 2 * time_test, 2 * time_test, 4 * time_test])
speed_4d = tangential_speed(signal_4d, time_test)
print("speed_4d =", speed_4d)                 # expected: [[5.], [5.], ..., [5.]]
assert np.allclose(speed_4d, 5)

print("All tests passed!")
```
