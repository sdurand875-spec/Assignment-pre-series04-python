# 4. Estimate the tangential speed of the signal

## 4.1 Tangential velocity vs tangential speed

- **Tangential velocity** is a **vector**: an arrow with a direction and a length. It tells us "I am going this way, at this rate." At each moment, this vector is always **tangent to the path** of the movement, which is why we call it *tangential* velocity.
- **Tangential speed** is a **scalar**: just a number, the length of that arrow. It only tells us "I am going at this rate," without any direction.

In other words, the tangential speed is the **magnitude** (or norm) of the tangential velocity vector:

$$\text{speed} = \|\vec{v}\|$$

This link is what we use in the next section to compute the tangential speed.

## 4.2 Tangential speed of a 3D movement

For a movement in 3D, the position at time $t$ is given by $x(t)$, $y(t)$ and $z(t)$. Their time derivatives $\dot{x}(t)$, $\dot{y}(t)$ and $\dot{z}(t)$ are the components of the velocity vector along each axis (computed in section 3):

$$\vec{v}(t) = \big(\dot{x}(t),\ \dot{y}(t),\ \dot{z}(t)\big)$$

The tangential speed is the magnitude (norm) of this vector. Using the Pythagorean theorem in 3D, we get:

$$v(t) = \|\vec{v}(t)\| = \sqrt{\dot{x}(t)^2 + \dot{y}(t)^2 + \dot{z}(t)^2}$$

The result is one positive number at each time $t$.

**Reference:** Ling, S. J., Sanny, J., & Moebs, W. (2016). *University Physics Volume 1*. OpenStax, section 4.1, Displacement and Velocity Vectors: https://openstax.org/books/university-physics-volume-1/pages/4-1-displacement-and-velocity-vectors
